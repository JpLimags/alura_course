```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
imigration_data = pd.read_csv(r'C:\Users\Jp Lima\Desktop\alura_course\datasets\canadian_immegration_data.csv')
imigration_data.head()
```

```python
imigration_data.info()
```

```python
imigration_data.set_index('Country', inplace=True)
```

```python
anos = list(map(str, range(1980, 2014)))
```

```python
brasil = imigration_data.loc['Brazil', anos]
brasil
```

```python
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}
brasil_data = pd.DataFrame(brasil_dict)
```

```python
brasil_data
```

### Criando os gráficos únicos e multiplos

```python
plt.figure(figsize=(8,4))
plt.plot(brasil_data['ano'], brasil_data['imigrantes'])
plt.title('Distribuição de imigrantes Brasileiros para o Canadá')
plt.xticks(['1980', '1985', '1990', '1995','2000', '2005', '2010'])
plt.xlabel('Ano')
plt.ylabel('Número de Imigrantes')
plt.plot()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(brasil_data['ano'], brasil_data['imigrantes'])
ax.set_title('Distribuição de imigrantes Brasileiros para o Canadá')
ax.set_xlabel('Anos')
ax.set_ylabel('Número de Imigrantes')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
fig, axs = plt.subplots(1,2, figsize=(15,5))

axs[0].plot(brasil_data['ano'], brasil_data['imigrantes'])
axs[0].set_title('Distribuição de imigrantes Brasileiros para o Canadá')
axs[0].set_xlabel('Anos')
axs[0].set_ylabel('Número de Imigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid(True)

axs[1].boxplot(brasil_data['imigrantes'])
axs[1].set_title('Boxplot dos imigrantes Brasileiros para o Canadá')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('Número de Imigrantes')
axs[1].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[1].grid(True)
```
