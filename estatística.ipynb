### Conclusões do caso abaixo

-  Conseguimos ver que apesar dos filmes terem realidades diferentes por utilizarmos um métrica inadequada acabamos tendo uma mesma pontuação mas que não nos mostra a realidade dos dados;

- Como podemos ver no gráfico a distribuição desses dados são completamente diferentes com isso a medida mais correta seria calcularmos o desvio padrão;

```python
filme_1 = [2.5]*10 + [3.5]*10
np.mean(filme_1)
```

```python
filme_2 = [5.0]*10 + [1]*10
np.mean(filme_2)
```

```python
plt.boxplot(x= [filme_1 , filme_2])
```

### Desvio padrão

```python
print(f'Filme 1: {np.std(filme_1)}\nFilme2: {np.std(filme_2)}')
```

```python
dados = np.array([[100, 120, 90, 110, 130], [80, 150, 70, 140, 60]])

df = pd.DataFrame(dados, index = ["Semana 1", "Semana 2"],
                  columns= ["Segunda", "Terça", "Quarta", "Quinta", "Sexta"])


desvio_padrao_semana1 = df.loc["Semana 1"].std()
desvio_padrao_semana2 = df.loc["Semana 2"].std()

print(df)
print("\nDesvio Padrão da Semana 1:", round(desvio_padrao_semana1, 2))
print("Desvio Padrão da Semana 2:", round(desvio_padrao_semana2, 2))
```
